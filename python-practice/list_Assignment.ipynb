### Assignment 6: List Methods

Create a list of random numbers and sort it in ascending and descending order. Remove the duplicates from the list and print the modified list.

```python
import random

random_list = [random.randint(-1, 1) for _ in range(10)]

print(random_list)

temp_acc_list = sorted(random_list)
temp_dec_list = sorted(random_list, reverse=True)

# remove duplicates from acc list 
prev1 = temp_acc_list[0]
prev2 = temp_dec_list[0]

acc_random_list = [prev1,]
dec_random_list = [prev2,]

for index in range(1, 10):
    n1: int = temp_acc_list[index]
    n2: int = temp_dec_list[index]

    if n1 != prev1:
        acc_random_list.append(n1)
        prev1 = n1

    if n2 != prev2:
        dec_random_list.append(n2)
        prev2 = n2

print(acc_random_list, dec_random_list)
```

### Assignment 14: List Rotation

Write a function that rotates a list by n positions. Print the original and rotated lists.



```python
#rotate the given list by 'k' places
def rotate_list(list, k):
    n: int = len(list)
    return [list[(n - k + i) % n] for i in range(0, n)]


rotate_list(int_list, 5)
```

### Assignment 15: List Intersection

Write a function that takes two lists and returns a new list containing only the elements that are present in both lists. Print the intersected list.

```python
list1 = [1, 2, 3, 4, 5, 6, 7]
list2 = [2, 4, 6, 8, 10]

# considering numbers don't repeat in either list
def merge_common_ele(list1, list2):
    return [num for num in list1 if list2.count(num) > 0]

merge_common_ele(list1, list2)
```
