### Assignment 1: Handling Division by Zero

Write a function that takes two integers as input and returns their division. Use try, except, and finally blocks to handle division by zero and print an appropriate message.

```python
def division(a: float, b: float) -> float | None:
    try:
        return a / b
    except ZeroDivisionError:
        print("Can't divide b zero!")
    finally:
        print("Function executed.")


num = int(input("Enter value for `a`"))
b = int(input("Enter value for `b`"))

division(num, b)
```

### Assignment 8: Exception Handling in Network Operations

Write a function that attempts to open a URL and read its contents. Use try, except, and finally blocks to handle network-related errors and print an appropriate message.

```python
import requests


def read_url(url):
    try:
        response = requests.get(url)
        response.raise_for_status()
        return response.text
    except requests.RequestException as e:
        print(f"Network error: {e}")
        return None
    finally:
        print("Execution complete.")
```

### Assignment 9: Exception Handling in JSON Parsing

Write a function that attempts to parse a JSON string. Use try, except, and finally blocks to handle JSONDecodeError if the string is not a valid JSON and print an appropriate message.

```python
import json


def parse_json(input: str):
    try:
        data = json.loads(input)
        return data
    except json.JSONDecodeError as e:
        print(e)
        return None
    finally:
        print("Execution completed")
```

### Assignment 10: Custom Exception Handling

Define a custom exception named `NegativeNumberError`. Write a function that raises this exception if a negative number is encountered in a list. Use try, except, and finally blocks to handle the custom exception and print an appropriate message.

```python
class NegativeNumberError(Exception):
    pass


def check_negative(num: int):
    try:
        if num < 0:
            raise NegativeNumberError(f"Negative number found: {num}")
        return num
    except NegativeNumberError as error:
        print(error)
    finally:
        print("Check completed")


check_negative(-1)
```

### Assignment 11: Exception Handling in Function Calls

Write a function that calls another function which may raise an exception. Use try, except, and finally blocks to handle the exception and print an appropriate message.

```python
def division_by_zero(dividend: int, divisor: int) -> float:
    if divisor == 0:
        raise ZeroDivisionError

    result = dividend / divisor
    return result


def caller(divisor: int) -> float | None:
    try:
        return division_by_zero(1, divisor)
    except ZeroDivisionError:
        print("Can't divide by zero.")
    finally:
        print("Division completed")


caller(0)
```

### Assignment 12: Exception Handling in Class Methods

Define a class with a method that performs a division operation. Use try, except, and finally blocks within the method to handle division by zero and print an appropriate message.

```python
class DivideNumbers:
    def __init__(self, dividend: int, divisor: int):
        self.dividend = dividend
        self.divisor = divisor

    def division(self) -> float | None:
        try:
            return self.dividend / self.divisor
        except ZeroDivisionError:
            print("Can't divide by zero.")
        finally:
            print("Division attempted.")


DivideNumbers(1, 4).division()
```

### Assignment 13: Exception Handling in Data Conversion

Write a function that takes a list of strings and converts them to integers. Use try, except, and finally blocks to handle ValueError if a string cannot be converted and print an appropriate message.

```python
def str_to_int(nums: list) -> list | None:
    try:
        return list(map(lambda s: int(s), nums))
    except ValueError:
        print("Can't convert to integer.")
    finally:
        print("Conversion attempted.")

str_to_int(["1", "2", "3", "a"])
```

### Assignment 14: Exception Handling in List Comprehensions

Write a function that uses a list comprehension to convert a list of strings to integers. Use try, except, and finally blocks within the list comprehension to handle ValueError and print an appropriate message.

```python
def str_to_int(nums: list) -> list | None:
    try:
        return [int(item) for item in nums]
    except ValueError:
        print("Can't convert to integer.")
    finally:
        print("Conversion attempted.")

str_to_int(["1", "2", "3", 4])
```

### Assignment 15: Exception Handling in File Writing

Write a function that attempts to write a list of strings to a file. Use try, except, and finally blocks to handle IOError and ensure the file is properly closed.

```python
list_of_string = ["a", "b", "c", "d", "e"]

try:
    with open("file_handling/test1.txt", "w") as file:
        content = ""
        for item in list_of_string:
            content += item + "\n"
        file.write(content)
except IOError as e:
    print(f"Error writing to file: {e}")
finally:
    print("Write attempt complete.")
```
