### Assignment 1: Fibonacci Sequence with Memoization
 
Define a recursive function to calculate the nth Fibonacci number using memoization. Test the function with different inputs.

```python
def nth_fibonacci(n, k = 1, prev = 0):
    return k if n == 1 else nth_fibonacci(n - 1, prev + k, k) 
    
print(nth_fibonacci(6))    
```

### Assignment 6: Function with Decorators

Define a function that calculates the time taken to execute another function. Apply this decorator to a function that performs a complex calculation. Test the decorated function with different inputs.

```python
import time
from typing import Any, Callable 

def timer(func: Callable[..., Any]) -> Callable[..., Any]: 
    def wrapper(*args: list, **kwargs: dict):
        time_start = time.time()
        result = func(*args, **kwargs)
        time_end = time.time()
        print(f"Time taken: {time_end - time_start}")
        return result
    return wrapper

@timer
def caller(delay: int):
    """Pause execution for a given amount of time.

    Args:
        delay: Time to wait, in seconds (e.g. 2 means 2 seconds).
    """
    time.sleep(delay)
    return

caller(2)
```

### Assignment 7: Higher-Order Function for Filtering and Mapping

Define a higher-order function that takes two functions, a filter function and a map function, along with a list of integers. The higher-order function should first filter the integers using the filter function and then apply the map function to the filtered integers. Test with different filter and map functions.

```python
def higher_order(func1: Callable[..., Any], func2, integer_list: list) -> list:
    filter_list = filter(func1, integer_list)
    mapped_list = map(func2, filter_list)
    return list(mapped_list)

higher_order(
    func1=lambda num: num % 2 == 0, 
    func2=lambda num: num**2, 
    integer_list=[1,2,3,4,5,6,7,8,9]
)
```

### Assignment 8: Function Composition

Define a function that composes two functions, f and g, such that the result is f(g(x)). Test with different functions f and g.

```python
def compose(f: Callable[..., Any], g: Callable[..., Any]) -> Callable[..., Any]:
    return lambda x: f(g(x))


f = lambda x: x**3
g = lambda x: x + 2
h = compose(f, g)
print(h(3))
```

### Assignment 10: Function with Error Handling

Define a function that takes a list of integers and returns their average. The function should handle any errors that occur (e.g., empty list) and return None in such cases. Test with different inputs.

```python
def average(int_lst: list) -> float | None:
    try:
        return sum(int_lst) / len(int_lst)
    except ZeroDivisionError:
        return None

# Test
print(average([1, 2, 3, 4, 5]))  # 3.0
print(average([]))  # None
```

### Assignment 11: Function with Generators

Define a function that generates an infinite sequence of Fibonacci numbers. Test by printing the first 10 numbers in the sequence.

```python
from functools import lru_cache


@lru_cache
def fibonacci(n: int) -> int:
    if n == 0:
        return 0
    if n == 1: 
        return 1
    return fibonacci(n - 1) + fibonacci(n - 2)

fibonacci(10)
```

### Assignment 12: Currying

Define a curried function that takes three arguments, one at a time, and returns their product. Test the function by providing arguments one at a time.

```python
def curry_product(x: int) -> Callable[..., Any]:
    def inner_1(y: int) -> Callable[..., Any]:
        def inner_2(z: int) -> int:
            return x * y * z

        return inner_2

    return inner_1


curry_product(8)(2)(3)
```
