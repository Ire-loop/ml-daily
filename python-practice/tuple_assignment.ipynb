### Assignment 15: Tuple Functions

Write functions that take a tuple and return the minimum, maximum, and sum of the elements. Print the results for a sample tuple.

```python
def max_in_tuple(tuple: tuple) -> int:
    max = float("-inf")

    for num in tuple:
        if num > max:
            max = num
    return int(max)


def min_in_tuple(tuple: tuple) -> int:
    min = float("inf")

    for num in tuple:
        if num < min:
            min = num
    return int(min)


def sum_tuple(tuple: tuple) -> int:
    sum = 0
    for num in tuple:
        sum += num
    return sum


print(max_in_tuple(pos_int), min_in_tuple(pos_int), sum_tuple(pos_int))
```
