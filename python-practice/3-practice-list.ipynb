```python
# Defining the list
fruits = ["apple", "banana", "cherry", "guava"]
```

```python
# Adding an element to end 
fruits.append("kivi")
fruits
```

```python
# Removing the element from last
fruits.pop()
fruits
```

```python
# Add an element to a specific index
fruits.insert(1, "watermelon")
fruits
```

```python
# Duplicate elements can be added 
fruits.append("watermelon")
fruits.insert(4, "watermelon")
fruits
```

```python
# Remove the first occurence of the element
fruits.remove("watermelon")
fruits
```

```python
# Index of first occurence of the element
fruits.index("watermelon")
```

```python
# Slice the array 
fruits[1:3]
```

```python
# Define sequence of the array 
fruits[1::2]
```

```python
# Slice and define the sequence
print(fruits)
print(fruits[1:4:-1]) # results in empty array
print(fruits[4:1:-2]) # seq. ommit 2 elements in reverse order
```

```python
# comprehensions
numbers = [1, 2, 3, 4, 5, 6, 7, 8]

odd_numbers = [num for num in numbers if num % 2 != 0] # syntax expression iteration condition(optional)
odd_numbers
```
