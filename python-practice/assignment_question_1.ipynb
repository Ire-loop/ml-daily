**Question 17:** Write a Python program to check if a year is a leap year.

```python
# Your code here
year = int(input())

if year % 100 == 0:
    if year % 400 == 0:
        print("Leap Year")
    else:
        print("Not a Leap Year")
elif year % 4 == 0:
    print("Leap Year")
else:
    print("Not a Leap Year")
```
