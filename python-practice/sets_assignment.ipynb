### Assignment 3: Set Operations

Create two sets: one with the first 5 positive integers and another with the first 5 even integers. Perform and print the results of union, intersection, difference, and symmetric difference operations on these sets.

```python
pos_int_set = set(num for num in range(1, 6))
pos_even_set = set(num * 2 for num in range(1, 6))

print(pos_int_set, pos_even_set, "\n")

print("Union", pos_int_set.union(pos_even_set))
print("Intersection", pos_int_set.intersection(pos_even_set))
print("Integer set difference with even set", pos_int_set.difference(pos_even_set))
print("Even set difference with integer set", pos_even_set.difference(pos_int_set))
print("Symmetric difference", pos_int_set.symmetric_difference(pos_even_set))
```
