### Assignment 2: Understanding Log Levels

Write a markdown cell that explains, in one sentence each, when you would use `DEBUG`, `INFO`, `WARNING`, `ERROR` and `CRITICAL`. Then list the five levels in order from least to most severe along with their numeric values (use `logging.DEBUG`, `logging.INFO`, etc. to print them).

- **DEBUG** – detailed diagnostic info used while developing, such as function arguments and intermediate values.
- **INFO** – confirms things are working as expected, e.g. a file was loaded or a connection established.
- **WARNING** – something unexpected happened but the program can continue, e.g. a divisor of 0 or a retried network call.
- **ERROR** – a specific operation failed but the app keeps running, e.g. inside an `except` block.
- **CRITICAL** – a serious failure where the app can't continue, e.g. a required database or config file is missing.

### Assignment 10: Logging Errors in a Real Function

Write a function `divide(a, b)` that returns `a / b`. Use `try`/`except` to catch `ZeroDivisionError`, log the problem at the `ERROR` level (try `logging.exception` as well and compare the output), and return `None`.

Log an `INFO` message when the division succeeds. Call the function once with valid values and once with `b = 0`, then inspect the log file.

```python
def divide(a: int, b: int) -> float | None:
    try:
        logging.info("Dividing...")
        result = a // b
    except ZeroDivisionError as ex:
        logging.exception(ex)
        return None
    else:
        logging.info("Division succeeded.")
        return result


print(divide(1, 2))
divide(1, 0)
```
