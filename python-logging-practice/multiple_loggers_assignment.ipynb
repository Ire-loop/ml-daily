### Assignment 9: Putting It All Together

Build a small application with three loggers: `app.auth`, `app.database` and `app.payment`.

- `app.auth` should show everything from `DEBUG` upward.
- `app.database` should show only `INFO` and above.

```python
import logging

auth_logger = logging.getLogger("app.auth")
auth_logger.setLevel(logging.DEBUG)

database_logger = logging.getLogger("app.database")
database_logger.setLevel(logging.INFO)

payment_logger = logging.getLogger("app.payment")
payment_logger.setLevel(logging.WARNING)

def login():
    auth_logger.debug("Logging...")

def fetch_user():
    database_logger.info("Fetching user...")

def charge():
    payment_logger.warning("Charging...")
```

```python
login()
fetch_user()
charge()
```
