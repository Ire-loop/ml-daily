### Assignment 1: Setup and Messy Dataset

Import `pandas` as `pd` and `numpy` as `np`. Then run the cell below. It creates a small sales dataset with **deliberately missing values** and saves it as `data.csv` (the lecture used a similar file). You will use `data.csv` for Parts 1-5.

The columns are `date`, `category`, `value`, `product`, `sales`, and `region`.

```python
import pandas as pd
import numpy as np

rng = np.random.default_rng(42)
n_rows = 30

messy = pd.DataFrame(
    {
        "date": pd.date_range("2026-01-01", periods=n_rows).strftime("%Y-%m-%d"),
        "category": rng.choice(["A", "B", "C"], n_rows),
        "value": rng.integers(10, 100, n_rows).astype(float),
        "product": rng.choice(["Product1", "Product2", "Product3"], n_rows),
        "sales": rng.integers(100, 1000, n_rows).astype(float),
        "region": rng.choice(["East", "North", "South", "West"], n_rows),
    }
)

# punch holes: 3 missing in `value`, 4 missing in `sales`
messy.loc[rng.choice(n_rows, 3, replace=False), "value"] = np.nan
messy.loc[rng.choice(n_rows, 4, replace=False), "sales"] = np.nan

messy.to_csv("data.csv", index=False)
```
