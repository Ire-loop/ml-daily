### Assignment 6: Multiple Plots with Subplot (2 x 2)

Using the data from Assignment 5, create a 2 x 2 grid of subplots:

1. `x` vs `y1` (green)
2. `y1` vs `x` (orange)
3. `x` vs `y2` (blue)
4. `x` vs `x` squared (red), where the squared values are computed with a list comprehension

Give each plot a title and call `plt.tight_layout()` before `plt.show()`.

```python
plt.subplot(2, 2, 1)
plt.plot(x, y, color="green", linestyle="--", marker="o", linewidth=2, markersize=10)
plt.title("Plot 1")

plt.subplot(2, 2, 2)
plt.plot(y, x, color="orange", linestyle=":", marker="o", linewidth=2, markersize=10)
plt.title("Plot 2")

plt.subplot(2, 2, 3)
plt.plot(x, z, color="blue", linestyle=":", marker="o", linewidth=2, markersize=10)
plt.title("Plot 3")

plt.subplot(2, 2, 4)
plt.plot(x, [num ** 2 for num in x], color="red", linestyle=":", marker="o", linewidth=2, markersize=10)
plt.title("Plot 4")

plt.tight_layout()
plt.show()
```

### Assignment 19: Revenue by Region and Payment Method

Using `sales_data`:

### Assignment 21: Dashboard and Saving a Figure

Build a 2 x 2 dashboard figure (`figsize=(14, 10)`) of the sales data with a bar chart (revenue by category), a line plot (revenue over time), a pie chart (revenue by region) and a histogram of `Total Revenue`. Add an overall title with `plt.suptitle`. Save the figure to `sales_dashboard.png` using `plt.savefig("sales_dashboard.png", dpi=150, bbox_inches="tight")` before calling `plt.show()`.

```python
plt.figure(figsize=(14, 10))

plt.subplot(2, 2, 1)
sales_data.groupby("Product Category")["Total Revenue"].sum().plot(kind="bar")
plt.title("Revenue by Product Category")

plt.subplot(2, 2, 2)
plt.plot(sales_data["Date"], sales_data["Total Revenue"], color="yellow", marker=None)
plt.xticks(rotation=45)
plt.title("Total Revenue by Date")

plt.subplot(2, 2, 3)
sales_data.groupby("Region")["Total Revenue"].sum().plot(kind="pie", autopct="%1.1f%%")
plt.title("Revenue by Region")

plt.subplot(2, 2, 4)
sales_data["Total Revenue"].plot(kind="hist", bins=100, color="red", edgecolor="black", linewidth=1)
plt.tight_layout()
```
