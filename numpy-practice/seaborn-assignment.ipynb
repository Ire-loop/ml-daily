### Assignment 13: Total Revenue by Product Category

Using `sales_df`, create a figure of size `(10, 6)` and draw a bar plot of **total** `Total Revenue` by `Product Category` using `estimator=sum`. Add the title "Total Sales by Product Category", an x-label, and a y-label. Rotate the x-tick labels by 45 degrees so they don't overlap. Which category earns the most revenue?

```python
plt.figure(figsize=(10, 5))
sns.barplot(x="Product Category", y="Total Revenue", data=sales_df, estimator=sum)
plt.title("Total Sales by Product Category")
plt.show()
```

### Assignment 14: Total Revenue by Region (Sorted)

Create the same kind of plot as Assignment 13, but for `Region`. This time, sort the bars from highest to lowest revenue. Calculate the order with pandas `groupby` and pass it to the `order` parameter. Add a title and axis labels. Which region is in first place and which is in last place?

```python
region_order = sales_df.groupby("Region")["Total Revenue"].sum().sort_values(ascending=False).index
region_order
```

```python

plt.figure(figsize=(10, 5))
sns.barplot(x="Region", y="Total Revenue", data=sales_df, estimator=sum, order=region_order)
plt.title("Total Sales by Product Category")
plt.show()
```

### Assignment 16: Revenue Trend Over Time

Convert the `Date` column of `sales_df` to datetime and create a new column `Month` that holds the month number. Create a line plot of total `Total Revenue` by `Month` using `estimator='sum'`, with `marker='o'`. Add a title and axis labels. Which month had the highest revenue?

### Assignment 17: Sales Dashboard

Combine what you have learned into one figure. Create a 2x2 grid of subplots with `figsize=(14, 10)` and fill it with: (1) a bar plot of total revenue by `Product Category`, (2) a box plot of `Total Revenue` by `Region`, (3) a histogram of `Total Revenue` with a KDE curve, and (4) a correlation heatmap of `Units Sold`, `Unit Price`, and `Total Revenue`. Give each subplot a title, add an overall title with `fig.suptitle`, and call `plt.tight_layout()` so nothing overlaps.

```python
plt.figure(figsize=(14, 10))

plt.suptitle("Sales Dashboard")

plt.subplot(2, 2, 1)
sns.barplot(x="Product Category", y="Total Revenue", data=sales_df, color="green")
plt.xticks(rotation=45)

plt.subplot(2, 2, 2)
sns.boxplot(x="Region", y="Total Revenue", data=sales_df, color="darkblue")
plt.xticks(rotation=45)

plt.subplot(2, 2, 3)
sns.histplot(x="Total Revenue", data=sales_df, kde=True, color="yellow")

plt.subplot(2, 2, 4)
sns.heatmap(sales_df[["Units Sold", "Unit Price", "Total Revenue"]].corr(), annot=True, cmap="coolwarm")

plt.tight_layout()
plt.show()
```
