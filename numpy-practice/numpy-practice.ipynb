### Assignment 13: Normalization (Standardization)

Create the array `[10, 20, 30, 40, 50]`. Normalize it so that it has a mean of `0` and a standard deviation of `1`, using the formula:

`normalized = (data - mean) / std`

Use `np.mean` and `np.std`. Print the original data, the mean, the standard deviation, and the normalized data. Finally, verify the result by printing the mean and standard deviation of the normalized array (they should be approximately `0` and `1`).

````python
data = np.array([10, 20, 30, 40, 50])

mean = np.mean(data)
std = np.std(data)

normalized = (data - mean) / std

print("Data", data)
print("Mean", mean)
print("Standard deviation", std)
print("Normalized Data", normalized)

print("Mean of Normalized Data", np.mean(normalized))
print("Standard deviation of Normalized Data", np.std(normalized))
````

Create a 2D array of marks for 5 students (rows) in 3 subjects (columns):

```
[[78, 85, 90],

```

Using only NumPy operations (no loops):

1. Print the average marks of each student (row-wise) and the average marks of each subject (column-wise).

3. Add `5` grace marks to every score in the third subject (last column) and print the updated array.
4. Normalize each subject (column) so that it has a mean of `0` and a standard deviation of `1`, and print the result.

````python
student_marks = np.array(
    [[78, 85, 90], [62, 70, 58], [95, 88, 92], [45, 52, 60], [80, 75, 83]]
)

shape = np.shape(student_marks)
print(shape)
````

````python
# Row wise average 
for row in range(shape[0]):
    print(f"Row {row + 1} average", np.mean(student_marks[row, ]))
````

````python
# Column wise average
for col in range(shape[1]):
    print(f"Column {col + 1} average", np.mean(student_marks[::, col]))
````

````python
student_marks[student_marks > 75]
````

````python
student_marks[::, 2] += 5
````
